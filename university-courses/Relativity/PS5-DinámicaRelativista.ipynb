```python
import numpy as np
import matplotlib.pyplot as plt
```

### Caso 1:

$$ \frac{v}{c^2} E_e' > P' \Longrightarrow P_{x_{centro}} > a \Longrightarrow P_x > 0$$

En este caso el neutrón se mueve con una velocidad alta y comparable con la de la luz.

El momentum en dirección $x$ siempre tomará una dirección positiva respecto al laboratorio, independientemente de la dirección que adquiera en el sistema de referencia propio del neutrón. Si bien en este último sistema la emisión puede ser isotrópica, en el laboratorio, el efecto relativista causa que las direcciones se inclinen hacia adelante en dirección del movimiento del neutrón (esto fue visto en el beaming relativista).

El valor mínimo que puede adquirir el momentum lineal horizontal es de:

$$ P_{x_{mín}} = P_{x_{centro}} - a = \gamma \left( \frac{v}{c^2} E_e' - P' \right) > 0 $$

Como el electrón siempre se mueve hacia adelante visto desde el laboratorio, el momentum que adquiere el electrón es mayor que su energía en reposo.

\

### Caso 2:

$$ \frac{v}{c^2} E_e' = P' \Longrightarrow P_{x_{centro}} = a \Longrightarrow P_x \geq 0 $$


La elipse es tangente al eje $P_y$ (respecto al laboratorio) y por tanto existe la posibilidad de que el momentum lineal sea cero. En este caso la velocidad del neutrón y la energía del electrón se sincronizan para alcanzar la magnitud del momentum lineal del electrón, de modo que el efecto relativista del movimiento del neutrón justo es contrarrestado por el momentum del electrón, respecto al laboratorio. Y respecto al neutrón, en este punto el electrón adquiere una dirección negativa.

\

### Caso 3:

$$ \frac{v}{c^2} E_e' < P' \Longrightarrow P_{x_{centro}} < a \Longrightarrow P_x < 0 $$

En este caso la elipse cruza el eje $P_y$, por lo que el momentum lineal horizontal puede tener valores negativos.

Visto por el neutrón, el electrón sale emitido con gran velocidad hacia direcciones negativas. Esto también se refleja en el sistema de referencia del laboratorio, donde el electrón contrarresta el efecto relativista del movimiento del neutrón.

En este caso el neutrón se mueve a bajas velocidades.

```python
def elipse(v, Ep, Pp, label):
    gamma = 1/np.sqrt(1 - v**2)     # Unidades canónicas
    Pcentro = gamma*v*Ep            # [P] = kg-luz

    a = gamma*Pp   # Semieje mayor
    b = Pp         # Semieje menor

    # Ángulo para parametrizar la elipse
    theta = np.linspace(0, 2*np.pi, 100)

    # Ecuaciones paramétricas de la elipse
    Px = Pcentro + a*np.cos(theta)
    Py = b*np.sin(theta)

    plt.plot(Px, Py, label=label)
    plt.scatter([Pcentro], [0])   # Centro de la elipse
    # plt.annotate('C', xy=(Pcentro, 0), xytext=(Pcentro, b/2), arrowprops=dict(facecolor='black', shrink=0.02))

# Caso 1: (v E')/c^2 > p'
v1 = 0.9    # c
Ep1 = 1.0
Pp1 = 0.5

fig = plt.figure()
elipse(v1, Ep1, Pp1, r'Caso 1: $\frac{v E^\prime}{c^2} > p^\prime$')

# Caso 2: (v E')/c^2 = p'
v2 = 0.8
Ep2 = 1.0
Pp2 = v2*Ep2
elipse(v2, Ep2, Pp2, r'Caso 2: $\frac{v E^\prime}{c^2} = p^\prime$')

# Caso 3: (v E')/c^2 < p'
v3 = 0.7
Ep3 = 1.0
Pp3 = 1.0
elipse(v3, Ep3, Pp3, r'Caso 3: $\frac{v E^\prime}{c^2} < p^\prime$')

plt.vlines(0, -1.3, 1.3, color='k', linestyle='--')
plt.hlines(0, -0.5, 3.3, color='k', linestyle='--')

# Configuraciones del gráfico
plt.xlabel(r'$P_x$ [kg-luz]')
plt.ylabel(r'$P_y$ [kg-luz]')
plt.title('Componentes del momentum lineal del electrón (Espacio de Fase)')
plt.legend()
plt.axis('equal')
plt.show()

fig.tight_layout()
```
