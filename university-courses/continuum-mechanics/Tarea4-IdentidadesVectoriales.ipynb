## Enunciado

**Objetivo**: El objetivo de esta tarea es repasar y demostrar algunas de las más importantes identidades vectoriales, y hacerlo utilizando la notación tensorial.

**Procedimiento**:

Demostrar las siguientes identidades vectoriales:

- $\nabla \cdot(f \mathbf{F})=f \nabla \cdot \mathbf{F}+\mathbf{F} \cdot \nabla f$

- $\nabla \times(f \mathbf{F})=f \nabla \times \mathbf{F}+\nabla f \times \mathbf{F}$

- $\nabla \cdot(\mathbf{F} \times \mathbf{G})=\mathbf{G} \cdot(\nabla \times \mathbf{F})-\mathbf{F} \cdot(\nabla \times \mathbf{G})$

- $\nabla \times(\nabla f)=\mathbf{0}$

- $\nabla \cdot(\nabla \times \mathbf{F})=0 $

- $\nabla^2(f g)=f \nabla^2 g+g \nabla^2 f+2 \nabla f \cdot \nabla g$

- $\nabla \cdot(\nabla f \times \nabla g)=0$

- $\nabla(\mathbf{F} \cdot \mathbf{G})=\mathbf{F} \times(\nabla \times \mathbf{G})+(\mathbf{F} \cdot \nabla) \mathbf{G}+\mathbf{G} \times(\nabla \times \mathbf{F})+(\mathbf{G} \cdot \nabla) \mathbf{F}$

- $\nabla \times(\mathbf{F} \times \mathbf{G})=(\mathbf{G} \cdot \nabla) \mathbf{F}+\mathbf{F}(\nabla \cdot \mathbf{G})-(\mathbf{F} \cdot \nabla) \mathbf{G}-\mathbf{G}(\nabla \cdot \mathbf{F})$

- $\nabla \times(\nabla \times \mathbf{F})=\nabla(\nabla \cdot \mathbf{F})-(\nabla \cdot \nabla) \mathbf{F}$

- $\frac{\partial(f \mathbf{F})}{\partial t} + \nabla \cdot(f \mathbf{F} \mathbf{F}) = f\frac{\partial \mathbf{F}}{\partial t} + f(\mathbf{F} \cdot \nabla) \mathbf{F}$

## Solución

**Nota:** Cada una de las demostraciones que se muestran a continuación son realizadas considerando un plano euclidiano en el cual los vectores de la base del sistema de coordenadas son ortonormales.

#### 1) $$\quad \nabla \cdot(f \mathbf{F})=f \nabla \cdot \mathbf{F}+\mathbf{F} \cdot \nabla f$$

\

$$
\begin{aligned}
\nabla \cdot(f \mathbf{F}) = \partial_i\left(f F^i\right) = (\partial_i f) F^i + f \partial_i F^i =\nabla F \cdot \mathbf{F}+ f\nabla \cdot \mathbf{F}
\end{aligned}
$$

\


#### 2) $$\quad \nabla \times(f \mathbf{F})=f \nabla \times \mathbf{F}+\nabla f \times \mathbf{F}$$

$$
\begin{aligned}
{[\nabla \times(f \mathbf{F})]^i } & = \epsilon_{j k}^i \partial^j\left(f F^k\right) = \epsilon_{j k}^i\left(\partial^j f\right) F^k + f \epsilon_{j k}^i \partial^j F^k \\
& = [(\nabla f) \times \mathbf{F}]^i + f (\nabla \times \mathbf{F})^i
\end{aligned}
$$

\


#### 3) $$\quad \nabla \cdot(\mathbf{F} \times \mathbf{G}) = \mathbf{G} \cdot(\nabla \times \mathbf{F})-\mathbf{F} \cdot(\nabla \times \mathbf{G})$$

$$
\begin{aligned}
\nabla \cdot(\mathbf{F} \times \mathbf{G}) & = \partial_i \left(\epsilon_{j k}^i F^j G^k \right) = \epsilon_{j k}^i \left(\partial_i F^j \right) G^k + \epsilon_{j k}^j F^j \left(\partial_i G^k \right) \\
& = G^k \epsilon_{j k}^i \partial_i F^j + F^j \epsilon_{j k}^i \partial_i G^k \\
\\
& \quad\quad \text{De acuerdo a las permutaciones: } \epsilon_{j k}^i = {{\epsilon_{k}}^i}_j \quad \text{y} \quad \epsilon_{j k}^i = - {{\epsilon_{j}}^i}_k \\
\\
& = G^k \left({{\epsilon_{k}}^i}_j \partial_i F^j \right) - F^j\left({{\epsilon_{j}}^i}_k \partial_i G^k\right) \\
& = G^k [\nabla \times \mathbf{F}]_k - F^j [\nabla \times \mathbf{G}]_j \\
& = \mathbf{G} \cdot(\nabla \times \mathbf{F})-\mathbf{F} \cdot(\nabla \times \mathbf{G})
\end{aligned}
$$

\


#### 4) $$\quad \nabla \times(\nabla f)=\mathbf{0}$$

\begin{aligned}
{[\nabla \times(\nabla f)]^i } & = \epsilon_{j k}^i \partial^j\left(\partial^k f\right) = \epsilon_{j k}^i \partial^k\left(\partial^j f\right) \\
& =-\epsilon_{k j}^i \partial^k\left(\partial^j f\right)=-[\nabla \times(\nabla F)]^i \\
\\
& \Longrightarrow \quad 2[\nabla \times(\nabla f)]^i = 0 \\
& \Longrightarrow \quad \nabla \times(\nabla f) = \mathbf{0}
\end{aligned}

\


#### 5) $$\quad \nabla \cdot(\nabla \times \mathbf{F}) = 0$$

$$
\begin{aligned}
\nabla \cdot(\nabla \times \mathbf{F}) & = \partial_i \left(\epsilon_{j k}^i \partial^j F^k \right) = \epsilon_{j k}^i \partial_i \partial^j F^k = \epsilon_{j k}^i \partial^j \partial_i F^k \\
& = -\partial^j \left({\epsilon_{j}}^i_k \partial_i F^k \right) = -\nabla \cdot(\nabla \times \mathbf{F}) \\
\\
& \Longrightarrow \quad 2 \nabla \cdot(\nabla \times \mathbf{F})=0 \\
& \Longrightarrow \quad \nabla \cdot(\nabla \times \mathbf{F})=0
\end{aligned}
$$

\


#### 6) $$\quad \nabla^2(f g) = f \nabla^2 g+g \nabla^2 f+2 \nabla f \cdot \nabla g$$

$$
\begin{aligned}
\nabla^2(f g) & = \partial_i \partial^i(f g) = \partial_i \left[g \partial^i f+ f \partial^i g \right] = \partial_i \left(g \partial^i f \right) + \partial_i \left(f \partial^i g \right) \\
& = \partial_i g \ \partial^i f + g \partial_i \partial^i f + \partial_i f \ \partial^i g + f \partial_i \partial^i g \\
& = \nabla g \cdot \nabla f + g \nabla^2 f + \nabla f \cdot \nabla g + f \nabla^2 g \\
& = f \nabla^2 g+g \nabla^2 f+2 \nabla f \cdot \nabla g
\end{aligned}
$$

\


#### 7) $$\quad \nabla \cdot(\nabla f \times \nabla g) = 0$$

$$
\begin{aligned}
& \text{Usando la identidad } (3): \\
& & \nabla \cdot(\nabla f \times \nabla g) & = \nabla q \cdot (\nabla \times \nabla f) - \nabla f \cdot (\nabla \times \nabla g) \\
& \text{Por la identidad } (4): \\
& & & = \nabla q \cdot \mathbf 0 - \nabla f \cdot \mathbf 0 = 0
\end{aligned}
$$

\


#### 8) $$\quad \nabla(\mathbf{F} \cdot \mathbf{G}) = \mathbf{F} \times(\nabla \times \mathbf{G})+(\mathbf{F} \cdot \nabla) \mathbf{G}+\mathbf{G} \times(\nabla \times \mathbf{F})+(\mathbf{G} \cdot \nabla) \mathbf{F}$$

Esta identidad se demuestra partiendo de la igualdad derecha de modo que se llega a la expresión izquierda.

$$
\begin{aligned}
& \left.[\mathbf{F} \times(\nabla \times \mathbf{G})]^i+[\mathbf{F} \cdot \nabla) \mathbf{G}\right]^i+[\mathbf{G} \times(\nabla \times \mathbf{F})]^i+[(\mathbf{G} \cdot \nabla) \mathbf{F}]^i \\
& =\epsilon_{j k}^i F^j(\nabla \times \mathbf{G})^k+F^j \partial_j G^i+\epsilon_{j k}^i G^j(\nabla \times \mathbf{F})^k+G^j \partial_j F^i \\
& =\epsilon_{j k}^i F^j \epsilon_m^{k l} \partial_l G^m+\epsilon_{j k}^i G^j \epsilon_m^{k l} \partial_l F^m+F^j \partial_j G^i+G^j \partial_j F^i \\
& =\epsilon_{j k}^i \epsilon_m^{k l}\left(F^j \partial_l G^m+G^j \partial_l F^m\right)+F^j \partial_j G^i+G^j \partial_j F^i \\
\\
& \quad\quad \text{Aplicamos la identidad:  } \epsilon_{j k}^i \epsilon_m^{k l}=\delta^{i l} \delta_{j m}-\delta_m^i \delta_j^l \\
\\
& = \delta^{i l} \delta_{j m}\left(F^j \partial_l G^m+G^j \partial_l F^m\right)-\delta_m^i \delta_j^l\left(F^j \partial_l G^m+G^j \partial_l F^m\right)+F^j \partial_j G^i+G^j \partial_j F^i \\
& = F^j \partial_i G^j+G^j \partial_j F^j \underbrace{-F^j \partial_j G^i} \overbrace{- G^j \partial_j F^i} + \underbrace{F^j \partial_j G^i} + \overbrace{G^j \partial_j F^i} \\
& = F^j \left(\partial_i G^j \right) + G^j \left(\partial_j F^j \right) = \partial_i \left(F^j G^j \right) \\
& =[\nabla(\mathbf{F} \cdot \mathbf{G})]^i
\end{aligned}
$$

\


#### 9) $$\quad \nabla \times(\mathbf{F} \times \mathbf{G})=(\mathbf{G} \cdot \nabla) \mathbf{F}+\mathbf{F}(\nabla \cdot \mathbf{G})-(\mathbf{F} \cdot \nabla) \mathbf{G}-\mathbf{G}(\nabla \cdot \mathbf{F})$$

$$
\begin{aligned}
\left[\nabla \times(\mathbf{F} \times \mathbf{G}) \right]^i & = \epsilon_{j k}^i \partial_j(\mathbf{F} \times \mathbf{G})^k = \epsilon_{j k}^i \partial_j \left(\epsilon_{i m}^k F^i G^m\right) \\
& =\epsilon_{j k}^i \epsilon_{l m}^k \partial_j\left(F^l G^m\right)=\epsilon_{j k}^i \epsilon_{L m}^k\left(F^l \partial_j G^m+G^m \partial_j F^l\right) \\
& =\left(\delta_l^i \delta_{j m}-\delta_m^i \delta_{j l}\right)\left(F^l \partial_j G^m+G^m \partial_j F^l\right) \\
& =F^i \partial_j G^j+G^j \partial_j F^i-F^j \partial_j G^i-G^i \partial_j F^j \\
& =[\mathbf{F}(\nabla \cdot \mathbf{G})]^i+[(\mathbf{G} \cdot \nabla) \mathbf{F}]^i-[(\mathbf{F} \cdot \nabla) \mathbf{G}]^i-[\mathbf{G}(\nabla \cdot \mathbf{F})]^i \\
& =[(\mathbf{G} \cdot \nabla) \mathbf{F}]^i+[\mathbf{F}(\nabla \cdot \mathbf{G})]^i-[(\mathbf{F} \cdot \nabla) \mathbf{G}]^i-[\mathbf{G}(\nabla \cdot \mathbf{F})]^i \\
&
\end{aligned}
$$

\


#### 10) $$\quad \nabla \times(\nabla \times \mathbf{F})=\nabla(\nabla \cdot \mathbf{F})-(\nabla \cdot \nabla) \mathbf{F}$$

\begin{aligned}
\left[\nabla \times(\nabla \times \mathbf{F}) \right]^i & = \epsilon^{i j}{ }_k \partial_j(\nabla \times \mathbf{F})^k = \epsilon_k^{i j} \partial_j \epsilon_{l m}^k \partial^l F^m \\
& =\left(\delta_l^i \delta_m^j-\delta_m^i \delta_l^j\right) \partial_j \partial^l F^m=\delta_l^i \delta_m^j \partial_j \partial^l F^m-\delta_m^i \delta_l^j \partial_j \partial^l F^m \\
& =\partial_j \partial^i F^j-\partial_j \partial^j F^i = \partial^i \left(\partial_j F^j \right) - \left(\partial_j \partial^j \right) F^i \\
& = [\nabla(\nabla \cdot \mathbf{F})]^i-[(\nabla \cdot \nabla) \mathbf{F}]^i
\end{aligned}

\


#### 11) $$\quad \frac{\partial(f \mathbf{F})}{\partial t} + \nabla \cdot(f \mathbf{F} \mathbf{F}) = f\frac{\partial \mathbf{F}}{\partial t} + f(\mathbf{F} \cdot \nabla) \mathbf{F}$$

\begin{aligned}
\left[ \frac{\partial(f \mathbf{F})}{\partial t} \right]^i + [\nabla \cdot(f \mathbf{F} \mathbf{F})]^i & = F^i \partial_t f + f \partial_t F^i+\partial_j \left(f F^j F^i \right) \\
& = \underbrace{F^i \partial_t f}+ f \partial_t F^i + \underbrace{F^i \partial_j\left(f F^j\right)}+f F^j\left(\partial_j F^i\right) \\
& =F^i\left[\partial_t+\partial_j \left(f F^j \right)\right] + f \left[\partial_t F^i+\left(F^j \partial_j \right) F^i\right] \\
& =F^i \left[\frac{\partial f}{\partial t} + \nabla \cdot(f \mathbf{F}) \right] + f \left[\frac{\partial \mathbf{F}}{\partial t}+(\mathbf{F} \cdot \nabla) \mathbf{F}\right]^i \\
& \quad\quad \text{Por la ecuación de continuidad:} \quad \frac{\partial f}{\partial t} + \nabla \cdot(f \mathbf{F})=0 \\
& =f\left[\frac{\partial \mathbf{F}}{\partial t}+(\mathbf{F} \cdot \nabla) \mathbf{F}\right]^i \\
\end{aligned}
