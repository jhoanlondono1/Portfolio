5. **La edad del universo**. Hay una forma de calcular la edad del universo, que en los experimentos anteriores tuvo que encontrarse manualmente.  Para ello se define el que se conoce como el *loopback time* (tiempo de mirada atrás), $t_{LB}$. Este número es el tiempo que se ha expandido el universo desde un valor específico del factor de escala $a(t)$ hasta el presente cuando $a(t_0)=1$.

   Partiendo de la definición del parámetro de Hubble:
   $$
   H=\frac{\dot a}{a}
   $$

   mostrar que el tiempo cósmico correspondiente a un determinado valor de $a$ se puede obtener como:

   $$
   t_{LB}(a) = t_0 - t(a) = \int_{a}^{1} \frac{\mathrm{d}a'}{a' H(a')}
   $$
   donde $t_0$ es el tiempo actual.

   Nótese que cuando $a=0$ el $t_{LB}$ es la edad $\tau_0$ del universo.

   Para calcular el $t_{LB}$ es necesario encontrar una ecuación que nos de $H$ como función de $a$. Eso se logra despejando $H$ de la ecuación de Friedman-Lemeitre:

   $$
   \epsilon=\frac{4 \pi}{3} G a^2\left[\frac{3 H^2}{8 \pi G}-\left(\rho+\rho_{\Lambda}\right)\right]
   $$
   donde recordemos $\epsilon$ es una constante.

   Demuestre que $H$ se puede escribir como:

   $$
   \frac{H^2}{H_0^2} = \frac{\rho+\rho_\Lambda + \rho_k}{\rho_{c0}}
   $$

   donde:
   $$
   \rho_k \equiv \frac{3 \epsilon}{4\pi G} \frac{1}{a^2}
   $$
   es una densidad ficticia que llamaremos *densidad de curvatura* (por un término usado en relatividad general, pero que aquí es solo una definición).

   Sabemos que la materia tiene una densidad que cambia según la ley $\rho_m=\rho_{0m}(a_0/a)^3$, que la energía oscura tiene $\rho_\Lambda = \rho_{0\Lambda}$ (no cambia en el tiempo) y que la densidad de curvatura varia según $\rho_k=\rho_{0k}(a_0/a)^2$. Usando esto mostrar que $H$ se puede escribir como:

   $$
   H(a) = H_0\sqrt{\Omega_{0m}(a_0/a)^3 + \Omega_{0k}(a_0/a)^2 + \Omega_\Lambda}
   $$

   En astrofísica, el corrimiento al rojo se relaciona con el factor de escala
   así: $a/a_0 = 1/(1+z)$.  Teniendo en cuenta la ecuación para $H(a)$ y escribiéndola en términos de $z$ el $t_{LB}$ es:

   $$
   t_{LB}(z) = t_H \int_0^z \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}}
   $$

   Y la edad del universo será:

   $$
   \tau_0 = t_H \int_0^\infty \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}}
   $$

   Donde $t_H=1/H_0$.

   Use la rutina `scipy.integrate.quad` para hacer esta integral (como limite superior puede poner `np.inf` y la rutina entiende que es una integral impropia) y calcular así la edad que tiene hoy el universo en los modelos cosmológicos estudiados en 3.

## Solución

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.integrate import solve_ivp, odeint, quad
```

### Punto 1

Vamos a introducir unas unidades convenientes para tratar numéricamente con problemas cosmológicos. En estas unidades $G=1$, $c=1$ y $H_0=1$. Llamaremos a estas las *unidades canónicas de la cosmología*. Determine el valor de las unidades de tiempo, longitud, masa y densidad.

\

**Unidad de tiempo:** Dado que la constante de Hubble tiene unidades inversas de tiempo, entonces:

$$H_0 = 1 \quad 1/U_T = 2.3*10^{-18} \quad 1/s ⟹ \boxed{U_T = \frac{1}{H_0} \quad s}$$

**Unidad de longitud:** Conociendo la unidad de tiempo, dado que la velocidad de la luz tiene unidades de longitud sobre tiempo, entonces:

$$c = 1 \quad U_L/U_T ⟹ \boxed{U_L = c U_T \quad m}$$

**Unidad de masa:** Esta se determina a partir de la constante de gravitación y sus unidades.

$$G = 1 \quad  \frac{U_L^3}{U_M U_T^2} ⟹ \boxed{U_M = \frac{U_L^3}{G U_T^2}  \quad kg}$$

\

Conociendo estas tres unidades canónicas base, es posible determinar las unidades canónicas de cantidades más complejas.

**Unidad de densidad:** Unidades de masa sobre volumen:

$$U_{\rho} = \frac{U_M}{U_L^3}$$

```python
H0 = 2.3e-18    # [1/s] - Constante de Hubble
G = 6.674e-11   # [m^3/(kg s^2)] - Constante de gravitación
c = 3.0e8       # [m/s] - Velocidad de la luz

""" Unidades canónicas de la cosmología """
UT = 1/H0
UL = c*UT
UM = UL**3/(G*UT**2)
Urho = UM/UL**3

print('Unidad canónica de densidad: ', Urho, ' [kg/m^3]')
```

### Punto 2

 Resuelva numéricamente las ecuaciones diferenciales para $H(t)$ y $\rho(t)$ derivadas de la conservación de la masa y la segunda ley de Newton (ecuación de Cauchy):

  $$
  \begin{aligned}
\dot{\rho} & =-3 \rho H(t) \\
\dot{H}+H^2 & =-\frac{4 \pi}{3} G \rho(t)
\end{aligned}
  $$

 Para ello asuma como valores actuales $\rho(t_0)=\rho_0=3\times 10^{-27}$ kg/m$^3$ y $H(t_0) = H_0 = 70$ km/s/Mpc. No nos interesa tanto el futuro como el pasado. Use $t_0=0$ y determine el valor $\tau$ (negativo, porque es en el pasado) para el cuál la integral produce valores muy grandes de la densidad (asíntota). Este valor, $\tau$ es la edad del universo predicha por el modelo. ¿Qué pasa con la edad del universo para valores de $\rho_0$ mayores? ¿Qué pasa para $\rho_0 = 0$ (universo vacío)?

```python
def Cauchy_cosmol(t,Y):
  rho, H = Y
  drhodt = -3*rho*H
  dHdt = -(4/3)*np.pi*rho - H**2
  #
  return np.array([drhodt, dHdt])

def densidad(rho0):
  # ============= Solución a las ecuaciones de Cauchy ===========
  Y0 = [rho0, 1]

  solp = solve_ivp(Cauchy_cosmol, [0,-1.2], Y0)
  solf = solve_ivp(Cauchy_cosmol, [0, 1], Y0)

  tsp = solp.t
  rhosp = solp.y[0]     # Densidad cosmológicas
  Hsp = solp.y[1]       # Parámetro de Hubble

  tsf = solf.t
  rhosf = solf.y[0]
  Hsf = solf.y[1]

  # ============ Edad del universo ============
  tau = tsp[rhosp == max(rhosp)][0]
  Myear = 1000000*365*24*3600     # [s] - Millón de años
  UTyear = UT/Myear
  edad = 13700/UTyear             # Edad "real" del universo

  print('================ EDAD DEL UNIVERSO ================')
  print('***** Densidad inicial: ρ = %.3e [kg  /m³] *****' %(rho0*Urho))
  print('\nModelo Universo fluido:\t %.3lf Millones de años' %(abs(tau)*UTyear))
  print('Edad "real": \t\t %.3lf Millones de años\n' %(edad*UTyear))

  # ============ Gráficos ============
  fig = plt.figure(figsize=(10, 5))

  plt.subplot(121)
  plt.title('Densidad - Universo-Fluido')
  plt.plot(tsp, rhosp, 'o--', label='Pasado', markersize=5)
  plt.plot(tsf, rhosf, 'o--', markersize=5, label='Futuro')
  plt.plot(0, rho0, 'o', label='Hoy')

  plt.vlines(-edad, -0.02, max(rhosp)+0.02, color='c', label='Edad real del universo')
  plt.vlines(tau, 0, max(rhosp), color='r', label='Edad modelada')

  plt.xlabel(f'Tiempo [$U_T =$ {UTyear:.3e} Maños]')
  plt.ylabel(f'Densidad: ρ [$U_ρ$ = {Urho:.3e} $kg/m^3$]')
  plt.legend()

  plt.subplot(122)
  plt.title('Parámetro de Hubble - Universo-fluido')
  plt.plot(tsp, Hsp, 'o--', markersize=5, label='Pasado')
  plt.plot(tsf, Hsf, 'o--', markersize=5, label='Futuro')
  plt.plot(0,1, 'o', label='Hoy')
  plt.xlabel(f'Tiempo [$U_T =$ {UTyear:.3e} Maños]')
  plt.ylabel('H [70 km/s/kPc]')
  plt.legend()

  fig.tight_layout()
#

rho0 = 3e-27        # [kg/m^3] - Densidad al tiempo de hoy
rho0 = rho0/Urho    # [UM/UL^3]
densidad(rho0)
```

* Notamos que integrando hacia atrás en el tiempo las ecuaciones de Cuachy para el universo fluido, la densidad cosmológica tiende a ser asintótica en u determinado tiempo, ese tiempo corresponde al nacimiento del universo, en donde toda la materia estaba condensada.

* El parámetro de Hubble aumenta hacia el pasado y hacia el futuro es prácticamente constante. Esto significa que cuando el universo estuvo en su época de muy extrema densidad ocurrió una expansión muy grande del espacio-tiempo pero pocos instantes de tiempo.

```python
densidad(0.0)
```

* Cuando la densidad inicial "hoy" es nula, hacia atrás en el tiempo la densidad no cambia, se mantiene constante, sin embargo, el parámetro de Hubble sí aumenta a medida que viajamos hacia el pasado.

```python
densidad(0.5)
```

* Notamos que si aumentamos la densidad cosmológica de "hoy", nos acercamos más a la etapa de mayor expansión y mayor densidad.

### Punto 3 - Solución a las ecuaciones de Friedmann-Lemaitre

Vamos ahora e estudiar la solución numérica, para distintas condiciones, las ecuaciones de Friedmann-Lemaitre:

  $$
\begin{aligned}
& \epsilon=\frac{4 \pi}{3} G a^2\left[\frac{3 H^2}{8 \pi G}-\left(\rho+\rho_{\Lambda}\right)\right] \\
& \ddot{a}=\frac{4 \pi}{3} G\left(2 \rho_{\Lambda}-\rho\right) a
\end{aligned}
$$
  
  Para ello integre numéricamente la ecuación de aceleración asumiendo distintos escenarios de materia.
  
  - $\Omega_0 = 0.1$, $\Omega_{0\Lambda} = 0$ (universo abierto sin energía oscura).
  - $\Omega_0 = 1$, $\Omega_{0\Lambda} = 0$ (universo crítico sin energía oscura).
  - $\Omega_0 = 2$, $\Omega_{0\Lambda} = 0$ (universo cerrado sin energía oscura).
  - $\Omega_0 = 1$, $\Omega_{0\Lambda} = 0.3$ (universo crítico sin baja energía oscura).
  - $\Omega_0 = 1$, $\Omega_{0\Lambda} = 0.7$ (universo crítico con alta energía oscura).
  - $\Omega_0 = 2$, $\Omega_{0\Lambda} = 0.7$ (universo cerrado con energía oscura).
  
  donde $\Omega_0 = \Omega_{0m} + \Omega_{0\Lambda} = (\rho_{0m} + \rho_{0\Lambda})/\rho_{0c}$ es el parámetro de densidad total.

  En cada escenario haga gráficos de la evolución de $a(t)$, $\ddot a(t)$, $\rho(t)$ y $H(t)$.

  Para ello se sugiere integrar la ecuación hacia adelante en el tiempo y hacia atrás en el tiempo por un tiempo $\tau$ primero pequeño y después aumentarlo hasta que se llegue a $a(t)=0$.

```python
def ecs_FL(t,Y, rho0o):
  a, ap, rho, H = Y

  # Integramos también las ecuaciones de Cauchy
  drhodt = -3*rho*H
  dHdt = -(4/3)*np.pi*rho - H**2

  dadt = H*a
  da2dt = (4*np.pi/3)*(2*rho0o - rho)*a

  return np.array([dadt, da2dt, drhodt, dHdt])

def fescala(om0, om0o, tf):
  #
  rho0c = 3/(8*np.pi)   # UM/UL^3 - Densidad crítica inicial
  om0m = om0 - om0o
  rho0 = om0m*rho0c     # Densidad inicial
  rho0o = om0o*rho0c    # Densidad de energía oscura inicial -> constante

  # ============ Solucionamos las ecuaciones de F-L ============
  Y0 = [1, 1, rho0, 1]
  te = np.linspace(0, -3, 100)
  solp = solve_ivp(ecs_FL, [0, te[-1]], Y0, t_eval=te, args=(rho0o,))

  # Edad del universo ------------------------------
  tsp = solp.t
  rhosp = solp.y[2]
  tau = tsp[rhosp == max(rhosp)][0]

  # Integración
  te = np.linspace(0, tf, 100)
  solf = solve_ivp(ecs_FL, [0, te[-1]], Y0, t_eval=te, args=(rho0o,))

  ts = np.concatenate((solf.t[::-1], solp.t))
  a = np.concatenate((solf.y[0][::-1], solp.y[0]))    # Factor de escala
  ap = np.concatenate((solf.y[1][::-1], solp.y[1]))   # Velocidad del factor de escala

  rho = np.concatenate((solf.y[2][::-1], solp.y[2]))
  H = np.concatenate((solf.y[3][::-1], solp.y[3]))

  app = 4*np.pi/3*(2*rho0o - rho)*a

  return a, app, rho, H, ts, tau
# ------------------------------------------------------------------------------
```

```python
# ============ Gráficos ============
Myear = 1000000*365*24*3600     # [s] - Millón de años
UTyear = UT/Myear

fig, axes = plt.subplots(1,4)
fig.set_size_inches(20,5)
fig.suptitle('Parámetros cosmológicos para diversos modelos de Universo', fontsize=15)

om0s = np.array([0.1, 1, 2, 1, 1, 2])
om0os = np.array([0.0, 0.0, 0.0, 0.3, 0.7, 0.7])

colores = ['b', 'g', 'r', 'orange', 'purple', 'k']
dic = {}
for i in range(len(om0s)):
  om0 = om0s[i]
  om0o = om0os[i]
  a, app, rho, H, ts, tau = fescala(om0, om0o, 4)

  # Almacenamos los resultados en un diccionario
  dic[i] = {'a':a, 'app':app, 'rho':rho, 'H':H, 'ts':ts, 'tau':tau}

  # Gráfico del factor de escala -----------------------
  c = colores[i]
  axes[0].plot(ts, a, color=c, label='$\Omega_0$ = %.1lf, $\Omega_{0Λ}$ = %.1lf' %(om0, om0o))
  axes[0].plot(0, 1, 'o', color=c)
  axes[0].vlines(tau, min(a), max(a), color=c, linestyles='dashed')

  # Gráfico de aceleración ------------------------------
  axes[1].plot(ts, app, color=c, label='$\Omega_0$ = %.1lf, $\Omega_{0Λ}$ = %.1lf' %(om0, om0o))
  axes[1].plot(0, 1, 'o', color=c)
  axes[1].vlines(tau, min(app), max(app), color=c, linestyles='dashed')

  # Gráfico de la densidad ------------------------------
  axes[2].plot(ts, rho, color=c, label='$\Omega_0$ = %.1lf, $\Omega_{0Λ}$ = %.1lf' %(om0, om0o))
  axes[2].plot(0, 1, 'o', color=c)
  axes[2].vlines(tau, min(rho), max(rho), color=c, linestyles='dashed')

  # Gráfico del parámetro de Hubble ---------------------
  axes[3].plot(ts, H, color=c, label='$\Omega_0$ = %.1lf, $\Omega_{0Λ}$ = %.1lf' %(om0, om0o))
  axes[3].plot(0, 1, 'o', color=c)
  axes[3].vlines(tau, min(H), max(H), color=c, linestyles='dashed')
#

axes[0].vlines(tau, min(a), max(a), color=c, label='τ', linestyles='dashed')
axes[0].set_title('Factor de escala')
axes[0].set_xlabel(f'Tiempo [$U_T =$ {UTyear:.3e} Maños]')
axes[0].set_ylabel('a(t)')
axes[0].legend(fontsize=9)

axes[1].vlines(tau, min(app), max(app), color=c, label='τ', linestyles='dashed')
axes[1].set_title('Aceleración del factor de escala')
axes[1].set_xlabel(f'Tiempo [$U_T =$ {UTyear:.3e} Maños]')
axes[1].set_ylabel('$\ddot{a}(t)$ [1/$U²_T$]')
axes[1].legend(fontsize=9)

axes[2].vlines(tau, min(rho), max(rho), color=c, label='τ', linestyles='dashed')
axes[2].set_title('Densidad')
axes[2].set_xlabel(f'Tiempo [$U_T =$ {UTyear:.3e} Maños]')
axes[2].set_ylabel(f'ρ(t)  [$U_ρ$ = {Urho:.3e} $kg/m^3$]')
axes[2].legend(fontsize=9)

axes[3].vlines(tau, min(H), max(H), color=c, label='τ', linestyles='dashed')
axes[3].set_title('Parámetro de Hubble')
axes[3].set_xlabel(f'Tiempo [$U_T =$ {UTyear:.3e} Maños]')
axes[3].set_ylabel('H(t)  [70 km/s/kPc]')
axes[3].legend(fontsize=9)

fig.tight_layout()
plt.show()
```

* De los gráficos notamos que el factor de escala disminuye hacia el pasado, teniendo a cero, y aumenta hacia el futuro. La disminución del factor de escala implica que el universo en sus primeros instantes de existencia tenía toda la materia extremadamente condensada, tanto que se podría considerar puntual.

* Luego del nacimiento el factor de escala comienza a acelerarse extremedamanete rápido hasta que comienza a suavizarse tendiendo a ser constante.

* Algo muy importante que observar es que para cuando un universo tienen mayor energía oscura tiende a acelerarse siempre en el futuro. Notemos que para un universo crítico con baja energía oscura podría ocurrir que la expansión muy lejos en el futuro frene y comience una contracción.

* En un universo cerrado sin energía oscura, podemos ntoar que en el futuro se logra una desaceleración de la expansión, comenzando una contracción.

### Punto 5 - La edad del universo

Hay una forma de calcular la edad del universo, que en los experimentos anteriores tuvo que encontrarse manualmente.  Para ello se define el que se conoce como el *loopback time* (tiempo de mirada atrás), $t_{LB}$. Este número es el tiempo que se ha expandido el universo desde un valor específico del factor de escala $a(t)$ hasta el presente cuando $a(t_0)=1$.

   Partiendo de la definición del parámetro de Hubble:
   $$
   H=\frac{\dot a}{a}
   $$

   mostrar que el tiempo cósmico correspondiente a un determinado valor de $a$ se puede obtener como:

   $$
   t_{LB}(a) = t_0 - t(a) = \int_{a}^{1} \frac{\mathrm{d}a'}{a' H(a')}
   $$
   donde $t_0$ es el tiempo actual.

   

> De acuerdo a la dependencia del parámetro de Hubble respecto al factor de escala, en la definición del parámetro es posible aplicar separación de variables, esto es:
>
> $\frac{da}{a H(a)} = dt$
>
> Integramos desde el punto $a(t)$ en el pasado hasta el presente en el punto $a(t_0) = 1$:
>
> $$\int_a^1 {\frac{da'}{a' H(a')}} = \int_t^{t_0} dt = t_0 - t(a)$$
>
> De acuerdo a la definición del *loopback time*:
>
> $$\boxed{t_{LB} = t_0 - t(a)} \quad (1)$$

Nótese que cuando $a=0$ el $t_{LB}$ es la edad $\tau_0$ del universo.

   Para calcular el $t_{LB}$ es necesario encontrar una ecuación que nos de $H$ como función de $a$. Eso se logra despejando $H$ de la ecuación de Friedman-Lemeitre:

   $$
   \epsilon=\frac{4 \pi}{3} G a^2\left[\frac{3 H^2}{8 \pi G}-\left(\rho+\rho_{\Lambda}\right)\right]
   $$
   donde recordemos $\epsilon$ es una constante.

   Demuestre que $H$ se puede escribir como:

   $$
   \frac{H^2}{H_0^2} = \frac{\rho+\rho_\Lambda + \rho_k}{\rho_{c0}}
   $$

   donde:
   $$
   \rho_k \equiv \frac{3 \epsilon}{4\pi G} \frac{1}{a^2}
   $$
   es una densidad ficticia que llamaremos *densidad de curvatura* (por un término usado en relatividad general, pero que aquí es solo una definición).

> Despejando $H^2$ de la ecuación de Friedman-Lemeitre, tenemos:
>
> $$H(a)^2 = \left( \frac{3 ϵ}{4 \pi G a^2} + \rho + \rho_{\Lambda} \right) \frac{8 \pi G}{3} \quad (2)$$
>
> La densidad crítica se define como $\rho_c = \frac{3 H(a)^2}{8 \pi G}$. Evaluando en $a(t_0) = 1$ y despejando tenemos:
>
> $$\frac{\rho_{c0}}{H_0^2} = \frac{3}{8 \pi G}$$
>
> Reemplazando la anterior ecuación en (2) y usando la definición de densidad de curvatura ($\rho_k$):
>
> $$H(a)^2 = \left( \rho_k + \rho + \rho_{\Lambda} \right) \frac{H_0^2}{\rho_{c0}}$$
>
> Despejando tenemos la forma funcional $H(a)$ como:
>
> $$\boxed{ \frac{H(a)^2}{H_0^2} = \frac{\rho+\rho_\Lambda + \rho_k}{\rho_{c0}} } \quad (3)$$

Sabemos que la materia tiene una densidad que cambia según la ley $\rho_m=\rho_{0m}(a_0/a)^3$, que la energía oscura tiene $\rho_\Lambda = \rho_{0\Lambda}$ (no cambia en el tiempo) y que la densidad de curvatura varía según $\rho_k=\rho_{0k}(a_0/a)^2$. Usando esto, mostrar que $H$ se puede escribir como:

   $$
   H(a) = H_0\sqrt{\Omega_{0m}(a_0/a)^3 + \Omega_{0k}(a_0/a)^2 + \Omega_\Lambda}
   $$

> Usando la ecuación $(2)$ y reemplazando las definiciones de $\rho = \rho_m$, $\rho_\Lambda$y $\rho_k$ tenemos:
>
> $$H(a) = H_0 \left[ \frac{\rho_{0m}}{\rho_{c0}} \left(\frac{a_0}{a}\right)^3 + \frac{\rho_{0\Lambda}}{\rho_{c0}} + \frac{\rho_{0k}}{\rho_{c0}} \left(\frac{a_0}{a}\right)^2 \right]^{1/2}$$
>
> Como el parámetro omega se define como, $\Omega = \rho/\rho_{c}$, evaluándolo en el tiempo $t_0$ tenemos:
>
> $$\Omega_{0m} = \rho_{0m}/\rho_{c0} \ , \quad \Omega_{0k} = \rho_{0k}/\rho_{c0} \ , \quad \Omega_{0\Lambda} = \rho_{0\Lambda}/\rho_{c0}$$
>
> Reemplazando en la ecuación anterior:
>
> $$\boxed{H(a) = H_0 \left[ \Omega_{0m} \left(\frac{a_0}{a}\right)^3 + \Omega_{0\Lambda} + \Omega_{0k} \left(\frac{a_0}{a}\right)^2 \right]^{1/2}} \quad (4)$$
>
> Demostramos entonces la identidad pedida.

En astrofísica, el corrimiento al rojo se relaciona con el factor de escala
   así: $a/a_0 = 1/(1+z)$.  Teniendo en cuenta la ecuación para $H(a)$ y escribiéndola en términos de $z$, el $t_{LB}$ es:

   $$
   t_{LB}(z) = t_H \int_0^z \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}}
   $$

   Y la edad del universo será:

   $$
   \tau_0 = t_H \int_0^\infty \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}}
   $$

   Donde $t_H=1/H_0$.

> Reemplazando la definición del factor de escala como función del corrimiento al rojo en la ecuación $(4)$, tenemos:
>
> $$H(a) = H_0 \left[ \Omega_{0m} (1+z)^3 + \Omega_{0\Lambda} + \Omega_{0k} (1+z)^2 \right]^{1/2} \quad (5)$$
>
> Derivando $a$ respecto a $z$: $$da = -\frac{a_0}{(1+z)^2} dz = -\frac{a}{1+z} dz$$
>
> Reemplazando la derivada anterior y la ecuación $(4)$ en la ecuación integral $(1)$ tenemos:
>
> $$t_{LB}(z) = t_0 - t(a) = -\frac{1}{H_0} \int_a^1 { \frac{dz}{ (1+z) \left[\Omega_{0m} (1+z)^3 + \Omega_{0\Lambda} + \Omega_{0k} (1+z)^2 \right]^{1/2} } }$$
>
> Definiendo $t_H = 1/H_0$, la ecuación se transforma en:
>
> $$\boxed{ t_{LB}(z) = t_H \int_0^z \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}} }$$
>
> Notemos que cuando $a ⟶ 0$, $t_{LB} ⟶ \tau_0$ (edad del universo), en este estado, $z ⟶ \infty$, por tanto la ecuación anterior se transforma en:
>
> $$\boxed{ \tau_0 = t_H \int_0^\infty \frac{dz}{(1+z)\sqrt{\Omega_{0m}(1+z)^3 + \Omega_{0k}(1+z)^2 + \Omega_\Lambda}} }$$
>
> La solución a la ecuación integral anterior corresponde a la edad de un universo con $\Omega_{0m}, \Omega_{0\Lambda}, \Omega_{0k}$.



   Use la rutina `scipy.integrate.quad` para hacer esta integral (como limite superior puede poner `np.inf` y la rutina entiende que es una integral impropia) y calcular así la edad que tiene hoy el universo en los modelos cosmológicos estudiados en 3.

```python
rho0c = 3/(8*np.pi)      # [UC] - Densidad crítica

edades = []
for i in range(len(om0s)):
  om0o = om0os[i]          # Omega de la energía oscura
  om0m = om0s[i] - om0o    # Omega de la materia

  e = 4*np.pi/3*rho0c*(1-(om0m + om0o))   # [UC] - Energía específica
  rho0k = 3*e/(4*np.pi)
  om0k = rho0k/rho0c

  integ = lambda z: 1/( (1+z)*(om0m*(1+z)**3 + om0k*(1+z)**2 + om0o)**0.5 )

  res, error = quad(integ, 0, np.inf)
  edades.append(res)
```

```python
print('============= Edad del universo "hoy" según cada modelo (Maños) =============\n')
print('Universo abierto sin energía oscura: \t τ = %.3e' %(abs(edades[0])*UTyear))
print('Universo crítico sin energía oscura: \t τ = %.3e' %(abs(edades[1])*UTyear))
print('Universo cerrado sin energía oscura: \t τ = %.3e' %(abs(edades[2])*UTyear))
print('\nUniverso crítico con baja energía oscura: \t τ = %.3e' %(abs(edades[3])*UTyear))
print('Universo crítico con alta energía oscura: \t τ = %.3e' %(abs(edades[4])*UTyear))
print('Universo cerrado con energía oscura: \t\t τ = %.3e' %(abs(edades[5])*UTyear))
```
