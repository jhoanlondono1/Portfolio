  $$
  \begin{aligned}
x & =R \sin \theta \cos \phi \\
y & =R \sin \theta \sin \phi \\
z & =R \cos \theta
\end{aligned}
  $$

  $$
  \mathrm{d} \ell^2=R^2\left(\mathrm{~d} \theta^2+\sin ^2 \theta \mathrm{~d} \phi^2\right)
  $$

  2. Calcule el escalar de curvatura para este espacio y exprese el radio como función de este escalar.

  \mathrm{d} V=\sqrt{g_{11}}\mathrm{~d} x^1 \sqrt{g_{22}}\mathrm{~d} x^n\cdots\sqrt{g_{nn}}\mathrm{~d} x^n$
  
    Usando esta expresión, calcule el volumen generalizado de la 2-esfera (que en su caso sería un área) y muestre que es igual a $4\pi R^2$.

```python
import numpy as np
import sympy as sp
import einsteinpy.symbolic as es
```

```python
'''-------------- Punto 1.4 : Escalar de curvatura para la 3-esfera --------------'''

# Definimos las variables simbólicas
R, chi, theta, phi = sp.symbols('R chi theta phi')

# Crea la métrica como una lista
g_list = [
    [R**2, 0, 0],
    [0, R**2*sp.sin(chi)**2, 0],
    [0, 0, R**2*sp.sin(chi)**2*sp.sin(theta)**2]
]

g_cov = es.MetricTensor(g_list, (R, chi, theta))
g_cov.tensor()
```

```python
g_contra = g_cov.change_config('uu')
g_contra.tensor()
```

El escalar de curvatura está dado por:

$$\rho = g^{\alpha \beta} R_{\alpha \beta} = g^{\alpha \beta} R^{\nu}_{\alpha \beta \nu}$$

Donde $R_{\alpha \beta}$ corresponde al tensor de Ricci.

```python
# Tensor de Riemann con primer índice contravariante
riemann = es.RiemannCurvatureTensor.from_metric(g_cov)
Rabcd = riemann.tensor()

print('Configuración del tensor: ', riemann.config)
Rabcd
```

```python
# Tensor de Ricci por contracción del Tensor de Riemann
ricci = es.RicciTensor.from_riemann(riemann)
Rab = ricci.tensor()
Rab
```

```python
g_contra_ = g_contra.tensor()

R_curvatura = sum(g_contra_[i, j]*Rab[i, j] for i in range(3) for j in range(3))

print("Escalar de Curvatura para la 3-esfera: \n")
R_curvatura
```

2. **Ecuaciones de Friedmann generales**. Partiendo de la métrica de Friedmann-Lemeaitre-Robertson-Walker:

   $$
   \mathrm{d} s^2=c^2 \mathrm{~d} t^2-a^2(t)\left[\frac{\mathrm{d} r^2}{1-K r^2}+r^2\left(\mathrm{~d} \theta^2+\sin ^2 \theta \mathrm{~d} \phi^2\right)\right]
   $$
   y de la ecuación de campo general:

   $$
   R_{\mu \nu}-\frac{1}{2} g_{\mu \nu} R=\frac{8 \pi G}{c^4} T_{\mu \nu} + \Lambda g_{\mu \nu}
   $$
   use el paquete `einsteinpy` para deducir las ecuaciones de Friedmann-Lemaitre generales:

   $$
   \frac{\ddot{a}}{a}=-\frac{4 \pi G}{3}\left(\rho+\frac{3 p}{c^2}\right)+\frac{\Lambda c^2}{3}
   $$
   $$
   \frac{\dot{a}^2+K c^2}{a^2}=\frac{8 \pi G \rho+\Lambda c^2}{3}
   $$

```python
t, r, theta, phi = sp.symbols('t r theta phi')       # Coordenadas
K, G, c, Lambda = sp.symbols('K G c Lambda')         # Constantes

a = sp.Function('a')(t)          # Factor de escala como función del tiempo
rho = sp.Function('rho')(t)      # Densidad de energía como función del tiempo
p = sp.Function('p')(t)          # Presión como función del tiempo

# Métrica Friedmann-Lemeaitre-Robertson-Walker como una matriz explícita
g = [
    [c**2, 0, 0, 0],
    [0, -a**2/(1 - K*r**2), 0, 0],
    [0, 0, -a**2*r**2, 0],
    [0, 0, 0, -a**2*r**2*sp.sin(theta)**2]
]

coords = [t, r, theta, phi]
g_ab = es.MetricTensor(g, coords)
g_ab_ = g_ab.tensor()
g_contra = g_ab.change_config('uu')
g_contra_ = g_contra.tensor()

# Tensor y escalar de Ricci (QUIZÁ NO SE REQUIERA)
R_ab = es.RicciTensor.from_metric(g_ab)
R_escalar = es.RicciScalar.from_metric(g_ab)

# Tensor de Einstein
G_ab = es.EinsteinTensor.from_metric(g_ab)
# G_ab = R_ab.tensor() - 1/2*g_ab_*R_escalar.expr
```

```python
G_ab_ = G_ab.tensor().simplify()

print('Tensor de Einstein: \n')
G_ab_
```

Para un **fluido perfecto**, el tensor de energía-impulso está dado por:

$$
T_{\mu\nu} = \left( \rho c^2 + p \right) U_\mu U_\nu - p g_{\mu\nu}
$$

Donde $U_\mu = (c, 0, 0, 0)$ para un observador comóvil.


```python
U_list = [c, 0, 0, 0]       # Observador comóvil
U = sp.Matrix(U_list)

T_list = [[(rho*c**2 + p)*U[i]*U[j] - p*g_ab_[i, j] for j in range(4)] for i in range(4)]
T_ab = es.StressEnergyMomentumTensor(T_list, (t, r, theta, phi))
T_ab_ = T_ab.tensor().simplify()

print('Tensor de energía-impulso para la métrica de FLRW: \n')
T_ab_
```

```python
# Ecuacion de campo general
eq_Einstein = G_ab_ - (8*sp.pi*G/c**4)*T_ab_ - Lambda*g_ab_

# Extraer ecuaciones relevantes
friedmann_00 = eq_Einstein[0, 0]
friedmann_11 = eq_Einstein[1, 1]

print("Ecuaciones de Friedmann: \n")
friedmann_00 = sp.Eq(friedmann_00, 0)
friedmann_00
```

```python
sp.expand(friedmann_11)
```

```python
dadt = sp.simplify(sp.Rational(8, 3)*sp.pi*G*a**2*rho - K*c**2 + sp.Rational(1, 3)*Lambda*c**2*a**2)

sp.expand(friedmann_11.subs({
    r : rho/a,
    sp.diff(a, t)**2 : dadt
}).simplify())
```
