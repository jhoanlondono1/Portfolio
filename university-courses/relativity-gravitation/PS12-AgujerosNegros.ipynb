```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
```

```python
c = 1       # Unidades luz
Rs = 1      # Unidad de distancia: radio de Schwarzschild

# Condiciones iniciales
t0 = 0
r0 = 1.01
r_max = 10*Rs

@interact(r0=(1.01,5,0.1))
def t_rayo_luz_saliendo(r0=1.01):
  r = np.linspace(r0,r_max,100)
  t_Sch = t0 + (r-r0)/c + Rs/c*np.log((r-Rs)/(r0-Rs))     # Espacio-tiempo curvo
  t_Min = t0 + (r-r0)/c     # Espacio-tiempo plano

  fig = plt.figure(figsize=(6,5))
  plt.title('Coordenada temporal para un rayo de luz emitido por un agujero negro')
  plt.plot(t_Sch, r, '--', label='Coord. Schwarzschild')
  plt.plot(t_Min, r, '--', label='Coord. Minkowski')
  plt.xlabel(r'$t$ $[U_L/R_S]$')
  plt.ylabel(r'$r$ $[R_S]$')
  plt.legend()
  fig.tight_layout()
```

**Observación:**

A medida que el rayo se aleja, el tiempo coordenado se incrementa más rápidamente de lo que cabría esperar en un espacio plano. Esto se debe a los efectos de la curvatura del espacio-tiempo cerca del agujero negro: la coordenada temporal de Schwarzschild “se estira” (dilatación gravitacional) al estar cerca del horizonte, por lo que para llegar a la misma $r_{Minkowski}$ hace falta más “tiempo coordenado” que en el caso plano.

```python
''' ------------------ Punto 1: Velocidad orbital ------------------ '''
# Constantes
pc = 3.0857e16        # [m] - Pársec
G = 6.6743e-11        # [N m^2/kg^2]
año = 365.25*24*3600  # [seg]
Msun = 1.98842e30     # [kg]
c = 299792458         # [m/seg]

# Unidades canónicas
UM = Msun
UL = pc
UT = np.sqrt(UL**3/(G*UM))
UV = UL/UT
c_can = c/UV

# Características de la estrella
alpha_mas = 125.38    # [mas] - Paralaje
alpha = 125.38e-3*np.pi/(60*60*180)     # [rad] - Paralaje
e = 0.88473           # Excentricidad
P_año = 16.0525       # [años] - Periodo
d_pc = 8122           # [pc] - Distancia

a = d_pc*alpha
P = P_año*año/UT
Mstar = (2*np.pi/P)**2*a**3
Rs = 2*Mstar/c_can**2
beta_max = Rs/(2*a)*(1+e)/(1-e)

print('Velocidad relativa máxima: %.5lf [c]' %np.sqrt(beta_max))
```

```python
''' ----------------- Punto 3: Corrimiento espectral ----------------- '''

t0 = 0          # Instante de paso en el periapsis -> La estrella parte del periapsis
n = 2*np.pi/P   # Velocidad angular relativa media
Es = np.linspace(0, 2*np.pi, 100)      # Anomalía excéntrica

t = lambda E: 1/n*(E - e*np.sin(E))    # Ecuación de Kepler
ts = t(Es)            # [UT] - Tiempos para cada E
ts_años = ts*UT/año   # [años]

r = lambda E: a*(1 - e*np.cos(E))      # Órbita elíptica
rs = r(Es)

betas = np.sqrt( Rs*(1/rs - 1/(2*a)) )   # Velocidad relativa

# Corrimiento espectral
z_ETcurvo = Rs/(2*rs) + betas**2/2
z_ETplano = betas**2/2

fig = plt.figure(figsize=(6,5))
plt.title('Corrimiento espectral en función del tiempo\n Movimiento kepleriano de la S0-2')
plt.plot(ts_años, z_ETcurvo, label='ET Curvo')
plt.plot(ts_años, z_ETplano, label='ET Plano')
plt.xlabel(r'$t$ [años]')
plt.ylabel(r'$z$')
plt.legend()
fig.tight_layout()
```

* En el periapsis ($t=0$ y $t=P$ años), cuando la estrella está más cerca del agujero negro, el corrimiento espectral debido a la curvatura del espacio-tiempo casi que duplica el corrimiento Doppler debido a la aberración de la luz en un espacio-tiempo plano. Esto es natural debido a que la luz no sólo sufre aberración relativista sino también por el potencial gravitatorio.

* Lejos del periapsis, cuando $t \approx 8$ años, ambas curvas tienden a valores muy próximos porque la velocidad de la estrella es más baja y la deformación del espacio-tiempo también es pequeña.

 1. Las coordenadas E-F se definen por la

 2. Usando el resultado 1 escriba la métrica Métrica de Schwarzschild en coordenadas de Eddington-Finkelstein (E-F):

   $$
   \mathrm{d} s^2=\left(1-\frac{R_{\mathrm{S}}}{r}\right) (c\mathrm{d} t^{\prime})^2-2 \frac{R_{\mathrm{S}}}{r} c \mathrm{~d} t^{\prime} \mathrm{d} r-\left(1+\frac{R_{\mathrm{S}}}{r}\right) \mathrm{d} r^2-\mathrm{d} \ell^2

  3. Demuestre que las geodésicas nulas radiales satisfacen la ecuación diferencial:

   $$

  4. Demuestre que esta ecuación diferencial es equivalente a dos ecuaciones, una para las geodesicas entrantes y otra para las salientes:

    \begin{aligned}
   \frac{dr}{dt'} & = -c\\

    haga gráficos $r-ct'$ de las geodésicas nulas que pasan por un determinado punto del espacio-tiempo, afuera, en y adentro del horizonte de eventos e interprete el resultado.

```python
''' Geodésicas nulas radiales afuera y adentro del horizonte de eventos '''

Rs = 1
c = 1
ctp0 = 0

# Geodésica entrante
r0_in = 2*Rs        # Punto exterior al horizonte
r_in_max = 0.1*Rs
r_in = np.linspace(r0_in, r_in_max, 100)

ctp_in = lambda r_in: ctp0 - (r_in - r0_in)
ctp_in_ = ctp_in(r_in)

# Geodésica saliente
r0_out = 1.1*Rs     # Punto exterior al horizonte
r_out_max = 2*Rs
r_out = np.linspace(r0_out, r_out_max, 100)

ctp_out = lambda r_out: ctp0 + r_out-r0_out + 2*Rs*np.log((r_out-Rs)/(r0_out-Rs))
ctp_out_ = ctp_out(r_out)

# Gráficos
fig = plt.figure(figsize=(6,5))
plt.title('Geodésicas nulas en coordenadas E-F.')
plt.plot(ctp_in_, r_in, '--', label='Rayo entrante')
plt.plot(ctp_out_, r_out, '--', label='Rayo saliente')
plt.hlines(1,0,max(ctp_out_), color='black', label='Horizonte de eventos')
plt.xlabel(r"$ct'$ $[R_s]$")
plt.ylabel(r"$r$ $[R_s]$")
plt.legend()
fig.tight_layout()
```

  $$
  K = R_{\mu\nu\alpha\beta} R^{\mu\nu\alpha\beta}
  $$
