3. **Efecto Doppler relativista**. En la óptica clásica la frecuencia de una onda $\nu$ que es observada proveniente desde una fuente que se mueve con velocidad $u_L$ está, por razones estrictamente cinemáticas y geométricas (que son comunes tanto en la relatividad newtoniana como en la relatividad especial), dada por:

   $$
   \nu = \frac{\nu_0}{1-u_L}
   $$
   donde $\nu_0$ es la frecuencia de la onda que sale de la fuente y que en la física clásica se asume igual a la frecuencia observada por ese sistema de referencia $\nu_0=\nu'$.

   **NOTA**: Este resultado corresponde a la frecuencia que percibe un detector que estando lejos de la fuente mira hacia -x, de modo que cuando $u_L>0$, es decir, cuando la fuente se mueve hacia +x, la frecuencia percibida por el detector es mayor (corrimiento al azul).

   Con base en este resultado:

   a. Muestre que, puesto que la duración de los fenómenos en el laboratorio y en la fuente no son los mismos debido a la dilatación del tiempo, la frecuencia observada en el sistema de referencia del laboratorio, se relaciona en definitiva con la frecuencia observada en el sistema de referencia de la fuente como:

   $$
   \nu = D\nu'  \\
   $$
   donde $D$ se conoce como el factor de Doppler y en este caso es igual a:

   $$
   D \equiv \sqrt\frac{1+u_L}{1-u_L}
   $$

   b. Demuestre que en el límite de bajas velocidades, $u\ll c$, las fórmulas anteriores se reducen a las fórmulas clásicas del efecto Doppler, a saber:

   $$
   z\equiv\frac{\Delta \lambda}{\lambda'} = \frac{u}{c}
   $$

   Donde $\Delta \lambda = \lambda-\lambda'$ y $z$ se conoce como el corrimiento espectral (informalmente, corrimiento al rojo).

   c. Haga un gráfico de la velocidad de la fuente $u_L$ en función del corrimiento al rojo $z$ para $z\in[0,14]$, que es el rango de los corrimientos observados en cosmología, usando la fórmula clásica y la fórmula relativista. ¿Qué concluye de este gráfico?.
   


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
c = 299792458

z = np.linspace(0, 14, 100)   # Rango de corrimientos observados en cosmología

ul_clas = lambda z: z
ul_rel = lambda z: (1 - (z+1)**2)/((z+1)**2 + 1)

plt.figure()
plt.title("Corrimiento al rojo ($z$)")
plt.xlabel("$z$")
plt.ylabel("$u_L$ [c]: Velocidad de alejamiento de la fuente")

plt.plot(z, abs(ul_rel(z)), label="Fórmula relativista", c='g')
plt.plot(z, ul_clas(z), label="Fórmula clásica", c='r')
plt.hlines(y=1, xmin=0, xmax=14, linestyle="--", label="Velocidad de la luz", color='orange')
plt.yscale("log")
plt.legend()
plt.show()
```

4. **Beaming relativistico y cambio en la intensidad de la luz de una fuente**. En clase probamos que en el beaming relativístico la relación entre el ángulo medido para una onda en el laboratorio y el medido en el sistema de referencia de la fuente es:

  $$
  \cos \theta=\frac{\cos \theta’+u_L}{1+u_L \cos \theta’}
  $$

  a. Si el ángulo sólido se define como $d\Omega = d(\cos\theta) d\phi$ muestre que:

  $$
  d\Omega = \frac{d\Omega'}{D^2}
  $$
  donde $D$ es el factor:

  $$
  D = \gamma(1+u_L\cos\theta')
  $$

  **NOTA**: Es interesante anotar que este factor coincide con el factor de Doppler $D$ visto en el punto 1 cuando $\theta'=0$ (el rayo va directamente hacia el observador). Esta no es una coincidencia. Más adelante mostraremos que el efecto Doppler que sufre una onda lumínica que se propaga en una dirección $\theta'$ en lugar de hacerlo directamente hacia el observador es:
  
  \begin{equation}
  \nu = D\nu' = \nu'\gamma(1+u_L\cos\theta')
  \tag{1}
  \end{equation}

  b. Si una fuente tiene intensidad específica $I'_\nu$ (potencia emitida por unidad de frecuencia por unidad de ángulo sólido), en el sistema de referencia de la fuente, la frecuencia a la que emite fotones será:

  $$
  \frac{I'_\nu d\nu'd\Omega'}{h\nu'}
  $$

  Suponiendo que esta frecuencia de emisión de fotones satisface la misma relación entre sistemas de referencia de la frecuencia ondulatoria (ver Ec. 1), muestre que la intensidad de la onda percibida es:
  
  $$
  I_\nu = D^3 I'_\nu
  $$

  c. Haga un gráfico de $D^3$ como función de $\theta$ para distintos valores de $u_L$ y comente.

$$D = \frac{1}{\gamma (1 - u_l \cos \theta)}$$

```python
uls = np.linspace(0.1, 0.99, 5)

plt.figure()
plt.title("Factor de Doppler cúbico en el Laboratorio")

for ul in uls:
  gamma = 1/np.sqrt(1 - ul**2)
  theta = np.linspace(0, np.pi/2, 100)
  D = 1/(gamma*(1 - ul*np.cos(theta)))

  plt.plot(theta*180/np.pi, D**3, label=r'$u = %.2f$ c' %ul)

plt.yscale("log")
plt.xlabel(r"$\theta$ [°]: Ángulo Observado")
plt.ylabel(r"$D(\theta)^3$")
plt.legend()
plt.show()
```
