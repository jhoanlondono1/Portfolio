### Punto c

**Tiempo en el laboratorio, $t(\tau)$:**

$$t(\tau) = \frac{1}{g_L} \sinh (g_L \tau)$$

**Velocidad, $v_L$:**

\begin{align}
  v_L(\tau) &= \tanh (g_L \tau) \\
  \\
  v_L(t) &= \frac{g_L t}{\sqrt{1+(g_L t)^2}}
\end{align}

**Posición, $x_L$:**

\begin{align}
  x_L(\tau) &= \frac{1}{g_L} [\cosh (g_L \tau) - 1] \\
  \\
  x_L(t) &= \frac{1}{g_L} (\sqrt{1+(g_L t)^2} - 1)
\end{align}

```python
""" ========================== Punto c ========================== """
c = 299792458   # [m/s] - Velocidad de la luz
g = 9.8         # [m/s^2] - Aceleración de la gravedad terrestre
año = 365.25*86400  # [s] - Año

gl = g/c*año    # [1/año] - Aceleración propia en la nave
print(f'Aceleración propia: g = 9.8 [m/s^2] <=> gL = {gl} [1/año]\n')

vl = lambda alphal, tau: np.tanh(alphal*tau)                  # Velocidad(τ)
xl = lambda alphal, tau: 1/alphal* (np.cosh(alphal*tau) - 1)  # [año] - Posición(τ)
t = lambda alphal, tau: 1/alphal* np.sinh(alphal*tau)         # [año] - Tiempo(τ)

vl_t = lambda alphal, t: gl*t/(np.sqrt(1+(gl*t)**2))          # [año/s] - Velocidad en el laboratorio
xl_t = lambda alphal, t: 1/alphal* (np.sqrt(1+(gl*t)**2) - 1) # [año] - Posición en el laboratorio

# Gráficos
mintau=0
maxtau=2.0
taus = np.linspace(mintau, maxtau, 50)       # [año] - Tiempo propio
ts = t(gl, taus)

fig, [axv, axx] = plt.subplots(1, 2, figsize=(12, 5))

# ========== Gráfico de velocidad ==========
# axv = fig.add_subplot(121)
axv.set_title("Velocidad de la nave")
axv.plot(taus, vl(gl, taus), c='r', label=r'$v_L(\tau)$')
axv.plot(ts, vl_t(gl, ts), c='g', label=r'$v_L(t)$')
axv.hlines(1, 0, max(ts), color='orange', label=r'Límite $c$', linestyles='--')

axv.plot([min(taus), max(taus)], [0,0], 'r->', linewidth=2, label=r'$\tau$ - Tiempo propio')
axv.legend()
axv.set_xlabel(fr"$t$ [año] - Tiempo en el laboratorio")
axv.set_ylabel(r"$v_L$")

# ========== Gráfico de posición ==========
# ax = fig.add_subplot(122)
axx.set_title("Trayectoria de la nave")
axx.plot(taus, xl(gl, taus), c='r', label=r'$x_L(\tau)$')
axx.plot(ts, xl_t(gl, ts), c='g', label=r'$x_L(t)$')
axx.plot(ts, ts, '--', c='black', label='Línea luminoide')

axx.plot([min(taus), max(taus)], [0,0], 'r->', linewidth=2, label=r'$\tau$ - Tiempo propio')
axx.legend()
axx.set_xlabel(r"$t$ [año] - Tiempo en el laboratorio")
axx.set_ylabel(r"$x_L$ [año-luz] - Posición")

fig.tight_layout()
```

**Observaciones:**

* De los gráficos podemos ver que a medida que el tiempo pasa y la nave acelera, la velocidad aumenta asintóticamente hacia la velocidad de la luz.

* También se puede ver la dilatación temporal que experimentan los observadores en la nave. El tiempo propio es mucho más pequeño que el tiempo medido en el laboratorio.

* Otra observación muy importante es que, cuando se describe la línea de universo ($x_L$) con el tiempo en el laboratorio, cuando la velocidad se acerca a la velocidad de la luz, esta línea tiende a ser paralela a la línea luminoide, que es la línea de universo de un fotón o una onda electromagnética  (línea negra), las cuales no tienen tiempo propio.

**Cuadrivelocidad:**

\begin{align}
  \tilde{U}_L(\tau) &: (\gamma, \sinh (g_L \tau)) = (\cosh (g_L \tau), \sinh (g_L \tau)) \\
  \\
  \tilde{U}_L(t) &: (\sqrt{1+(g_L t)^2}, g_L t)
\end{align}

\
**Cuadriaceleración:** $$\tilde{A}_L : g_L (U_{Lx}, U_{Lt})$$

```python
""" ======================= Diagramas espacio temporales ======================= """
# Cuadrivelocidad
Ul = lambda alphal, tau: np.array([np.cosh(alphal*tau), np.sinh(alphal*tau)])
Ul_t = lambda alphal, t: np.array([np.sqrt(1 + (alphal*t)**2), alphal*t])

# Cuadriaceleración
Al = lambda alphal, tau: alphal*Ul(alphal, tau)[::-1]
Al_t = lambda alphal, t: alphal*Ul_t(alphal, t)[::-1]

# ========== Gráficos espacio vs tiempo propio ==========
#
mintau=0
maxtau=2.0
taus = np.linspace(mintau, maxtau, 50)       # [año] - Tiempo propio

fig, [ax1, ax2] = plt.subplots(1, 2, figsize=(10, 5))

# ax = fig.add_subplot(121)
ax1.set_title("Trayectoria de la nave")
ax1.plot(taus, xl(gl, taus), c='limegreen', label='Línea de universo', zorder=1)
ax1.plot(taus, taus, '--', c='black', linewidth=1)

n = len(taus)
taus2 = [0, taus[n//3], taus[n//3*2]]

for taui in taus2:
  # Cuadrivelocidad
  Uli = Ul(gl, taui)
  x, y = taui, xl(gl, taui)
  dx, dy = Uli*0.2    # Escalamos la velocidad a 0.2
  Ul_arrow = ax1.arrow(x,y, dx, dy, width=0.01, head_length=0.04, head_width=0.05, fc='b', ec='b', zorder=3, label='Cuadrivelocidad')

  # Cuadriaceleración
  Ali = Al(gl, taui)
  x, y = taui, xl(gl, taui)
  dx, dy = Ali*0.2
  Al_arrow = ax1.arrow(x,y, dx, dy, width=0.01, head_length=0.04, head_width=0.05, fc='r', ec='r', zorder=3, label='Cuadriaceleración')

# Agregar leyenda con el label personalizado para las flechas
ax1.legend(handles=[ax1.lines[0], Ul_arrow, Al_arrow])
ax1.set_xlabel(r"$\tau$ [año] - Tiempo propio")
ax1.set_ylabel(r"$x_L$ [año-luz] - Posición")
ax1.grid()

# ========== Gráfico espacio vs tiempo en el laboratorio ==========
ts = t(gl, taus)

# ax = fig.add_subplot(122)
ax2.set_title("Trayectoria de la nave")
ax2.plot(ts, xl_t(gl, ts), c='limegreen', label='Línea de universo', zorder=1)
ax2.plot(ts, ts, '--', c='black', linewidth=1, label='Línea luminoide')

n = len(ts)
ts2 = [0, ts[n//3], ts[n//3*2]]

for ti in ts2:
  # Cuadrivelocidad
  Uli = Ul_t(gl, ti)
  x, y = ti, xl_t(gl, ti)
  dx, dy = Uli*0.5  # Escalamos la velocidad a la mitad
  Ul_arrow = ax2.arrow(x,y, dx, dy, width=0.01, head_length=0.04, head_width=0.05, fc='b', ec='b', zorder=3, label='Cuadrivelocidad')

  # Cuadriaceleración
  Ali = Al_t(gl, ti)
  x, y = ti, xl_t(gl, ti)
  dx, dy = Ali*0.5
  Al_arrow = ax2.arrow(x,y, dx, dy, width=0.01, head_length=0.04, head_width=0.05, fc='r', ec='r', zorder=3, label='Cuadriaceleración')

# Agregar leyenda con el label personalizado para las flechas
ax2.legend(handles=[ax2.lines[0], Ul_arrow, Al_arrow])
ax2.set_xlabel(r"$t$ [año] - Tiempo en el laboratorio")
ax2.set_ylabel(r"$x_L$ [año-luz] - Posición")
ax2.grid()

fig.tight_layout()
```

**Observaciones:**

* Podemos ver que en instante de tiempo cero, la cuadriaceleración y cuadrivelocidad son perpendicular. Esta ortogonalidad se sigue manteniendo si trasladamos la línea luminoide al vértice entre los vectores, pues es a través de esta que se define la perpendicularidad de ambos cuadrivectores.

4. **Leyes de transformación de la velocidad**. El cuadrivector de velocidad $\tilde{U_L}:\gamma(1,\vec{v}_L)$, como cualquier cuadrivector transforma siguiendo las leyes de transformación generales:

  $$
\begin{aligned}
& A^{0^{\prime}}=\gamma\left(A^0-\vec{u}_L \cdot \vec{A}_L\right) \\
& \vec{A}_L^{\prime}=\vec{A}_L+\left(\frac{\gamma-1}{u_L^2} \vec{u}_L \cdot \vec{A}_L-\gamma A^0\right) \vec{u}_L
\end{aligned}
  $$

  a. Use estas reglas generales y la definición de $\tilde{U_L}$ para demostrar que la transformación del factor de Lorentz es:

   $$
   \frac{\gamma_v}{\gamma_{v^{\prime}}}=\frac{D}{\gamma_u}
   $$
   donde

   $$
   D \equiv \frac{1}{1-\vec{u}_L \cdot \vec{v}_L}
   $$

  b. Use estas reglas generales y la definición de $\tilde{U_L}$ para demostrar que la transformación general de velocidades espaciales es:

    $$
    \vec{v}_L^{\prime}=\frac{D}{\gamma_u}\left[\vec{v}_L+\left(\frac{\gamma_u-1}{u_L^2} \vec{u}_L \cdot \vec{v}_L-\gamma_u\right) \vec{u}_L\right]
    $$

  c. Toda velocidad puede descomponerse como $\vec v_L : (v_{\|L},v_{\perp L})$ donde las cantidades en el par ordenado son las componentes de $\vec v_L$ que son paralela a $u_L$ y perpendicular a ella.  Muestre que, de la ecuación anterior, estas componentes pueden escribirse como:

   $$
   \begin{aligned}
v_{\| L}^{\prime} & =D\left(v_{\| L}-u_L\right) \\
v_{\perp L}^{\prime} & =\frac{D}{\gamma} v_{\perp L}
\end{aligned}
   $$
   y que estas coinciden con las transformaciones cuando se usa la configuración estándar.
