1. **El Lagrangiano de una carga eléctrica en el espacio-tiempo**. Los principios y ecuaciones básicas del formalismo Lagrangiano se que se cumplen en el espacio, se cumplen también en el espacio-tiempo. Por ejemplo, si una partícula tiene asociada una función lagrangiana $L(\{x^\mu\},\{{\dot x}^\mu\},\tau)$, donde aquí ${\dot x}^\mu = d{\dot x}^\mu/d\tau$ entonces la ecuación de movimiento estará dada por la versión generalizada de las ecuaciones de Euler-Lagrange (De la Torre, 2006):

$$
\frac{\partial L}{\partial x^\alpha}-\frac{d}{d \tau} \frac{\partial L}{\partial {\dot x}^\alpha}=0
$$

  Probar que, si se define la función lagrangiana de una carga eléctrica en un campo electromagnético con cuadripotencial $A^\mu$:
  $$
  L=-m_0 c \sqrt{\frac{d x^\mu}{d \tau} \frac{d x_\mu}{d \tau}}-\frac{q}{m_0 c} p^\mu A_\mu
  $$
  entonces la ecuación de movimiento es:

  $$
  \frac{d p^\alpha}{d \tau}=\frac{q}{c} F^{\alpha \beta} U_\beta
  $$
  que coincide exactamente la de una partícula con carga $q$ sometida a la fuerza de Lorentz.

2. **El lagrangiano del campo electromagnético**. Pruebe que si se supone que el funcional lagrangiano del campo electromagnético es:

  $$

  entonces las ecuaciones de Maxwell se pueden obtener a partir de la versión de las ecuaciones de Euler-Lagrange para funcionales con coordenadas continuas:

  $$
  \frac{\partial}{\partial x^\alpha} \frac{\partial ℒ}{\partial \omega_{,\alpha}}-\frac{\partial ℒ}{\partial \omega} = 0
  $$

  donde en general $ℒ (\omega,\{\omega_{,\mu}\})$.

4. **Transformación general del tensor de Faraday [Numérico]**. Usar herramientas de cálculo simbólico (`sympy`o `einsteinpy`) para encontrar la forma general, es decir, cuándo la velocidad $\vec u_L$ no está en la configuración estándar, en la que transforman los campos eléctrico y magnético, y demostrar que son equivalentes a las expresiones vectoriales:

\begin{aligned}
& \vec{E}=\vec{E}^{\prime}+(\gamma-1) \vec{E}_{\perp}^{\prime}-\gamma \vec{u}_L \times \vec{B}’ \\
& \vec{B}=\vec{B}^{\prime}+(\gamma-1) \vec{B}_{\perp}^{\prime}+\gamma \vec{u}_L \times \vec{E}^{\prime}
\end{aligned}

\

**Solución:**

Transformamos el tensor de Faraday a partir de la matriz de transformaciones generales de Lorentz-Einstein:

$$
\Lambda^{\mu'}_{\phantom{\mu}\mu} (\vec {u}_L) =
\begin{bmatrix}
\gamma & -\gamma u_{Lx} & -\gamma u_{Ly} & -\gamma u_{Lz} \\
-\gamma u_{Lx} & 1 + \alpha u_{Lx}^2 & \alpha u_{Lx} u_{Ly} & \alpha u_{Lx} u_{Lz} \\
-\gamma u_{Ly} & \alpha u_{Lx} u_{Ly} & 1 + \alpha u_{Ly}^2 & \alpha u_{Ly} u_{Lz} \\
-\gamma u_{Lz} & \alpha u_{Lx} u_{Lz} & \alpha u_{Ly} u_{Lz} & 1 + \alpha u_{Lz}^2
\end{bmatrix}
$$

donde $\alpha = \frac{\gamma-1}{u_L^2}$

$$F^{\mu' \nu'} = \Lambda^{\mu'}_\mu \Lambda^{\nu'}_\nu F^{\mu \nu}$$

En forma matricial es equivalente a $F' = \Lambda F \Lambda^T$.

```python
gamma, uLx, uLy, uLz = sp.symbols('gamma u_Lx u_Ly u_Lz')     # Variables en unidades luz
uL = sp.Matrix([uLx, uLy, uLz])     # Velocidad relativa
uL2 = (uL.T.dot(uL))

# Matriz de transformaciones generales del Lorentz-Einstein
alpha = (gamma-1)/uL2
Lambda = lambda uLx, uLy, uLz: sp.Matrix([
    [gamma, -gamma*uLx, -gamma*uLy, -gamma*uLz],
    [-gamma*uLx, 1+alpha*uLx**2, alpha*uLx*uLy, alpha*uLx*uLz],
    [-gamma*uLy, alpha*uLx*uLy, 1+alpha*uLy**2, alpha*uLy*uLz],
    [-gamma*uLz, alpha*uLx*uLz, alpha*uLy*uLz, 1+alpha*uLz**2]
])

# Del sistema S a S'
Lambda_uL = Lambda(uLx, uLy, uLz)

# Tensor de Faraday en el sistema S
Ex, Ey, Ez, Bx, By, Bz = sp.symbols('E_x E_y E_z B_x B_y B_z')
F = sp.Matrix([
    [0, -Ex, -Ey, -Ez],
    [Ex, 0, -Bz, By],
    [Ey, Bz, 0, -Bx],
    [Ez, -By, Bx, 0]
])

# Tensor transformado al sistema S'
F_p = sp.simplify(Lambda_uL*F*Lambda_uL.T)

# Separamos la componente del campo eléctrico y magnético en el sistema S'
Ex_p, Ey_p, Ez_p = F_p[1, 0], F_p[2, 0], F_p[3, 0]
Bx_p, By_p, Bz_p = F_p[3, 2], F_p[1, 3], F_p[2, 1]

Ep = sp.Matrix([Ex_p, Ey_p, Ez_p])
Bp = sp.Matrix([Bx_p, By_p, Bz_p])

# Campo eléctrico en el sistema S'
Ep
```

Para demostrar la equivalencia con las expresiones vectoriales, debemos descomponer los campos en componentes paralelas y perpendiculares a $\vec u_L$:

La componente paralela del campo eléctrico (o magnético) es la proyección a lo largo de la dirección del vector $\vec u_L$. Esto es:

$$\vec E'_{||} = (\vec E' \cdot \vec u_L) \hat u_L \quad ⟶ \quad \vec E_{\perp} = -\vec E'_{||} + \vec E'$$

Similarmente para el campo magnético.



```python
''' Equivalencia con las expresiones vectoriales en S'''
## Componentes paralelas a la dirección de movimiento de S'
Ep_ll = sp.simplify((Ep.T.dot(uL))*uL/uL2)
Bp_ll = sp.simplify((Bp.T.dot(uL))*uL/uL2)
Ep_l_ = sp.simplify(-Ep_ll + Ep)
Bp_l_ = sp.simplify(-Bp_ll + Bp)

E = sp.simplify(Ep + (gamma-1)*Ep_l_ - gamma*(uL.cross(Bp)))
B = sp.simplify(Bp + (gamma-1)*Bp_l_ + gamma*(uL.cross(Ep)))

sp.simplify(E)
```

* Vemos que de las componentes paralelas y perpendiculares de $\vec E'$ y $vec B'$ en las transformaciones vectoriales, obtenemos la expresión natural del campo eléctrico:

$$\vec E = [E_x, E_y, E_z]$$

5. **Tensor de Faraday en otros sistemas de coordenadas**. Las transformaciones de Lorentz son solo una de las posibles transformaciones que puede tener el tensor. Otro conjunto de transformaciones son las que corresponden a cambios de sistemas de coordenadas, por ejemplo de un sistema de coordenadas $x_\mu$ a otro $\bar {x}_\mu$. Como sabemos, en este caso el tensor transforma como:

  $$
F^{\bar\mu\bar\nu} = \frac{\partial \bar x^{\bar\mu}}{\partial x^{\mu}}\frac{\partial \bar x^{\bar\nu}}{\partial x^{\nu}} F^{\mu\nu}
$$
  Usando este hecho, escribir el tensor de Faraday, tanto $F^{\bar\mu\bar\nu}$ como $F_{\bar\mu\bar\nu}$ en coordenadas cilíndricas y en coordenadas esféricas, es decir, en términos de las componentes cilíndricas y esféricas de los campos.  Puede apoyarse de paquetes de cálculo simbólico para encontrar la respuesta.

  Como referencia, el tensor de Faraday, completamente covariante, en coordenadas esféricas es (ver https://arxiv.org/pdf/gr-qc/0409025):

  $$
  F_{\mu \nu}:
  \left(\begin{array}{cccc}
0 & E_r & r E_\theta & r \sin \theta E_\phi \\
-E_r & 0 & -r B_\phi & r \sin \theta B_\theta \\
-r E_\theta & r B_\phi & 0 & -r^2 \sin \theta B_r \\
-r \sin \theta E_\phi & -r \sin \theta B_\theta & r^2 \sin \theta B_r & 0
\end{array}\right)
  $$

  **Ayuda**: Pueden ser de utilidad las relaciones encontradas en esta página: https://en.wikipedia.org/wiki/Del_in_cylindrical_and_spherical_coordinates.

```python
t, x, y, z = sp.symbols('t x y z')
Ex, Ey, Ez, Bx, By, Bz = sp.symbols('E_x E_y E_z B_x B_y B_z')

# Forma contravariante del tensor de Faraday
F_contra = sp.Matrix([
    [0, -Ex, -Ey, -Ez],
    [Ex, 0, -Bz, By],
    [Ey, Bz, 0, -Bx],
    [Ez, -By, Bx, 0]
])
```

```python
''' --------------------- En componentes cilíndricas --------------------- '''
rho, phi, Erho, Ephi, Brho, Bphi = sp.symbols('rho phi E_rho E_phi B_rho B_phi')

coords_cart = sp.Matrix([t,x,y,z])

rho_ = sp.sqrt(x**2 + y**2)
phi_ = sp.atan2(y, x)
coords_cil = sp.Matrix([t,rho_,phi_,z])

# Jacobiano de cartesianas a cilíndricas
J_cart_cil = sp.simplify(coords_cil.jacobian(coords_cart))
# J_cart_cil = sp.Matrix([
#     [1, 0, 0, 0],
#     [0, sp.cos(phi), sp.sin(phi), 0],
#     [0, -rho*sp.sin(phi), rho*sp.cos(phi), 0],
#     [0, 0, 0, 1]
# ])

# Tensor contravariante de Faraday en el sistema S'
Fp_cil_xyz = J_cart_cil*F_contra*J_cart_cil.T

# Componentes cartesianas de los campos en términos de las cilíndricas
T_cil_cart = sp.Matrix([
    [sp.cos(phi), -sp.sin(phi), 0],
    [sp.sin(phi), sp.cos(phi), 0],
    [0, 0, 1]
])
E_cil = sp.Matrix([Erho, Ephi, Ez])
E_cil = sp.simplify(T_cil_cart*E_cil)

B_cil = sp.Matrix([Brho, Bphi, Bz])
B_cil = sp.simplify(T_cil_cart*B_cil)

Fp_cil = Fp_cil_xyz.subs({
    x: rho*sp.cos(phi),
    y: rho*sp.sin(phi),
    Ex: E_cil[0],
    Ey: E_cil[1],
    Ez: E_cil[2],
    Bx: B_cil[0],
    By: B_cil[1],
    Bz: B_cil[2]
})

print("Tensor de Faraday contravariante en coordenadas cilíndricas para el sistema S':\n")
sp.simplify(Fp_cil)
```

```python
''' --------------------- En componentes esféricas --------------------- '''
r, theta, phi = sp.symbols('r theta phi')
Er, Etheta, Br, Btheta = sp.symbols('E_r E_theta B_r B_theta')

r_ = sp.sqrt(x**2 + y**2 + z**2)
theta_ = sp.acos(z/r_)

coords_cart = sp.Matrix([t,x,y,z])
coords_sph = sp.Matrix([t,r_,theta_,phi_])

# Jacobiano de cartesianas a esféricas
J_cart_sph = sp.simplify(coords_sph.jacobian(coords_cart))
# J_cart_sph = sp.Matrix([
#     [1, 0, 0, 0],
#     [0, sp.sin(theta)*sp.cos(phi), r*sp.cos(theta)*sp.cos(phi), -r*sp.sin(theta)*sp.sin(phi)],
#     [0, sp.sin(theta)*sp.sin(phi), r*sp.cos(theta)*sp.sin(phi), r*sp.sin(theta)*sp.cos(phi)],
#     [0, sp.cos(theta), -r*sp.sin(theta), 0]
# ])

# Tensor contravariante de Faraday en el sistema S'
Fp_sph_xyz = J_cart_sph*F_contra*J_cart_sph.T

# Componentes cartesianas de los campos en términos de las esféricas
T_sph_cart = sp.Matrix([
    [sp.sin(theta)*sp.cos(phi), sp.cos(theta)*sp.cos(phi), -sp.sin(phi)],
    [sp.sin(theta)*sp.sin(phi), sp.cos(theta)*sp.sin(phi), sp.cos(phi)],
    [sp.cos(theta), -sp.sin(theta), 0]
])
E_sph = sp.Matrix([Er, Etheta, Ephi])
E_sph = sp.simplify(T_sph_cart*E_sph)

B_sph = sp.Matrix([Br, Btheta, Bphi])
B_sph = sp.simplify(T_sph_cart*B_sph)

Fp_sph = Fp_sph_xyz.subs({
    x: r*sp.sin(theta)*sp.cos(phi),
    y: r*sp.sin(theta)*sp.sin(phi),
    z: r*sp.cos(theta),
    Ex: E_sph[0],
    Ey: E_sph[1],
    Ez: E_sph[2],
    Bx: B_sph[0],
    By: B_sph[1],
    Bz: B_sph[2]
})

print("Tensor de Faraday contravariante en coordenadas esféricas para el sistema S':\n")
sp.simplify(Fp_sph)
```

Para la componentes covariantes utilizamos la métrica de Minkowski:

$$F_{\mu \nu} = \eta_{\mu \alpha} \eta_{\nu \beta} F^{\alpha \beta} \quad ⟶ \quad F' = \eta F \eta^T$$

```python
eta_cart = sp.diag(-1, 1, 1, 1)     # Métrica contravariante

# Transformamos la métrica a coordenadas cilíndricas
eta_cil = sp.simplify(J_cart_cil*eta_cart*J_cart_cil.T)
eta_cov_cil = eta_cil.inv()

# Transformamos la métrica a coordenadas esféricas
eta_sph = sp.simplify(J_cart_sph*eta_cart*J_cart_sph.T)
eta_cov_sph = eta_sph.inv()

# Tensor de Faraday covariante en el sistema S'
Fp_cov_cil = eta_cov_cil*Fp_cil*eta_cov_cil.T
Fp_cov_cil = Fp_cov_cil.subs({
    x: rho*sp.cos(phi),
    y: rho*sp.sin(phi),
    Ex: E_cil[0],
    Ey: E_cil[1],
    Ez: E_cil[2],
    Bx: B_cil[0],
    By: B_cil[1],
    Bz: B_cil[2]
})


print("Tensor de Faraday covariante en coordenadas cilíndricas para el sistema S':\n")
sp.simplify(Fp_cov_cil)
```

```python
Fp_cov_sph = eta_cov_sph*Fp_sph*eta_cov_sph.T
Fp_cov_sph = Fp_cov_sph.subs({
    x: r*sp.sin(theta)*sp.cos(phi),
    y: r*sp.sin(theta)*sp.sin(phi),
    z: r*sp.cos(theta),
    Ex: E_sph[0],
    Ey: E_sph[1],
    Ez: E_sph[2],
    Bx: B_sph[0],
    By: B_sph[1],
    Bz: B_sph[2]
})

print("\nTensor de Faraday covariante en coordenadas esféricas para el sistema S':\n")
sp.simplify(Fp_cov_sph)
```

6. **Tensor electromagnético dual [Numérico]**. Una cantidad de mucha utilidad en la electrodinámica es el tensor electromagnético dual, que se define como:

  $$
  {}^{*}F^{\mu\nu} = \frac{1}{2}\epsilon^{\mu\nu\alpha\beta}F_{\alpha\beta}
  $$

  a. Encontrar, usando `sympy` o `einsteinpy`, las componentes explícitas del dual contravariante ${}^{*}F^{\mu\nu}$ y el dual covariante ${}^{*}F_{\mu\nu}$

     **Ayuda**: Hay distintas maneras de hacerlo. Puede usar la rutina `sympy.LeviCivita` que permite calcular las componentes del tensor de Levi-Civita de cualquier dimensión.

  b. Usando el dual mostrar, usando `sympy` o `einsteinpy` que:

     $$
     (\vec E\cdot\vec B) = \frac{1}{4}{}^{*}F^{\mu\nu}F_{\mu\nu}
     $$

\
La forma completamente contravariante del tensor de Levi-Civita se obtiene multiplicando su forma covariante por la métrica de Minkowski $g_{\mu \nu}$.

$$
\epsilon_{\mu\nu\rho\sigma} = g_{\mu\alpha} g_{\nu\beta} g_{\rho\gamma} g_{\sigma\delta} \epsilon^{\alpha\beta\gamma\delta}
$$

Como el determinante de la métrica es -1, entonces:

$$ \epsilon_{\mu\nu\rho\sigma} = - \epsilon^{\mu\nu\rho\sigma} $$

Por tanto, el tensor dual electromagnético covariante tiene la forma:

 $$
  {}^{*}F_{\mu\nu} = -\frac{1}{2}\epsilon^{\mu\nu\alpha\beta}F^{\alpha\beta}
  $$

```python
''' Punto a: TENSOR DUAL ELECTROMAGNÉTICO '''

mu, nu, alpha, beta = sp.symbols('mu nu alpha beta')
Ex, Ey, Ez, Bx, By, Bz = sp.symbols('E_x E_y E_z B_x B_y B_z')

F_cov = sp.Matrix([         # Tensor de Faraday covariante
    [0, Ex, Ey, Ez],
    [-Ex, 0, -Bz, By],
    [-Ey, Bz, 0, -Bx],
    [-Ez, -By, Bx, 0]
])

F_contra = sp.Matrix([         # Tensor de Faraday covariante
    [0, -Ex, -Ey, -Ez],
    [Ex, 0, -Bz, By],
    [Ey, Bz, 0, -Bx],
    [Ez, -By, Bx, 0]
])

# Tensor dual contravariante *F^{\mu\nu}
F_dual_contra = sp.zeros(4, 4)
F_dual_cov = sp.zeros(4, 4)
for mu in range(4):
  for nu in range(4):
    for alpha in range(4):
      for beta in range(4):
        F_dual_contra[mu, nu] += 0.5*sp.LeviCivita(mu, nu, alpha, beta)*F_cov[alpha, beta]
        F_dual_cov[mu, nu] += -0.5*sp.LeviCivita(mu, nu, alpha, beta)*F_contra[alpha, beta]

print("Tensor dual contravariante: \n")
sp.simplify(F_dual_contra)
```
