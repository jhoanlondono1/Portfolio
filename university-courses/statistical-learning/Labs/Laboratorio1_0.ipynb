###2.
Renombrar las columnas en PascalCase Ejemplo: La columna radius_worst, concave points_se deberia llamar: RadiusWorst, ConcavePointsSe

```python
a = []

for col in df.columns:
    a.append(col.replace("_"," "))
df.columns = a

han = []
for c in df.columns:
    w = c.split() #el split solo funciona cuando hay espacios, por ende si no tengo espacios debo hacer un replace para cambiar "_" por " "
    z = []
    for i in w:
        z.append(i.capitalize())
    han.append("".join(z)) #join: concateneme con un "" en el medio(puede ser "_" o "/")
    
df.columns = han
df
```

El diagrama de caja es uno de los más útiles a la hora de analizar datos ya que de forma gráfica podemos observar su mediana, sus cuartiles y sus valores atípicos. Dentro de la caja obtenemos los valores dentro del cuartil 1 y 3, donde la mediana(cuartil 2), representado por la linea dentro de la caja, es la mediana. Los "brazos" que salen de su caja encierran los valores entre el máximo y mínimo no atípico de la distribución y los valores por fuera del diagrama representan los valores atípicos
